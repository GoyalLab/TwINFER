# Decomposing co-expression in twin sample space

TwINFER leverages information from recently divided sister cells, or “twins”, to infer gene regulatory networks: twin information discriminates regulatory from non-regulatory correlations and resolves interaction direction and type (activation/repression). Shortly after division, both twins belong to essentially the same state, and the initial *difference* in their gene expression levels is minimal. Across the population, by contrast, cell-state heterogeneity can correlate genes without regulation, for example two genes that are both highly expressed in one state and lowly expressed in the other.

This page derives the decomposition of the co-expression into two components, shared and unshared by the twins, following the Methods of the TwINFER paper, and relates it to the TwINFER pipeline in section 4.

## 1. Definitions

Consider a sample $S$ with cells expressing the two genes X and Y, with corresponding transcript counts $x_i(t)$ and $y_i(t)$ in the $i$-th cell at time $t$. The co-expression of genes X and Y is defined as

$$
\rho := \rho(x,y) = \operatorname*{spearman}_{i\in S}\left[x_i(t),\, y_i(t)\right].
\tag{E1}
$$

Now imagine that each cell in $S$ has divided to yield a pair of twin cells at $t = 0$ and let us define the ordered twin sample set $\hat{S}$, where each element $(i,j)\in\hat{S}$ is a twin pair $(i\neq j)$. Note that $(i,j)$ and $(j,i)$ are both in the set. After division the $N$ twin pairs represent $2N$ individual cells, for which we compute the ranks $r_{x_i}(t)$ and $r_{y_i}(t)$ of the $i$-th cell as before. Since twins inherit the same initial condition and dynamics, the co-expression in the offspring population is

$$
\operatorname*{spearman}_{(i,j)\in\hat{S}}\left[x_i(t),\, y_i(t)\right] = \operatorname*{spearman}_{i\in S}\left[x_i(t),\, y_i(t)\right] = \rho.
\tag{E2}
$$

The fact that we can think of $\rho$ in the twin sample space without modifications will later allow us to express the conventional co-expression in terms of components in twin sample space.

In the twin sample space $\hat{S}$, we define the twin shared correlation component

$$
\hat{\rho}_s := \hat{\rho}_s(x,y) = \operatorname*{spearman}_{(i,j)\in\hat{S}}\left[x_i(t),\, y_j(t)\right],
\tag{E3}
$$

namely, the correlation across twin pairs of the expression of gene X in one cell with the expression of gene Y in its twin. For convenience we introduce the notation

$$
\hat{h}_x := \hat{\rho}_s(x,x), \qquad \hat{h}_y := \hat{\rho}_s(y,y).
\tag{E4}
$$

Next we consider the difference $\Delta x = x_i - x_j$. We define the twin difference correlation

$$
\hat{\rho}_\Delta := \hat{\rho}_\Delta(\Delta x,\Delta y) = \operatorname*{pearson}_{(i,j)\in\hat{S}}\left[\Delta r_x(t),\, \Delta r_y(t)\right],
\tag{E5}
$$

where $r_{x_i}(t)$ and $r_{y_i}(t)$ are the ranks across the sample, and $\Delta r_x = r_{x_i} - r_{x_j}$. A subtle point to note is that this is a correlation between a *difference* of ranks and thus not a Spearman correlation.

**In TwINFER.** A hat denotes a quantity calculated between twins, and a $\Delta$ subscript a correlation between the *differences* in gene expression within the pair.

| symbol | meaning | after division |
|---|---|---|
| $\rho$ | ordinary gene-expression correlation across a population of single cells | — |
| $\hat{\rho}_s$ | correlation of gene X in one cell with gene Y in its twin | — |
| $\hat{h}_x$, $\hat{h}_y$ | correlation of each gene between twins | close to 1, then decrease as stochastic gene expression makes the trajectories of the twins diverge |
| $\hat{\rho}_\Delta$ | correlation of the differences $\Delta x$ and $\Delta y$ within twin pairs | $\Delta x \sim 0$ and $\Delta y \sim 0$ right after division |

## 2. Decomposition

We start by computing the correlation in Eq. (E5). The covariance is given by

$$
\begin{aligned}
\operatorname{cov}(\Delta x,\Delta y) &= \operatorname{cov}(x_i,y_i) + \operatorname{cov}(x_j,y_j) - \operatorname{cov}(x_i,y_j) - \operatorname{cov}(x_j,y_i)\\
&= 2\left[\operatorname{cov}(x_i,y_i) - \operatorname{cov}(x_i,y_j)\right],
\end{aligned}
\tag{E6}
$$

where for clarity we omitted the time dependence and wrote the equation for the counts (the ranks will be substituted at the end). Similarly,

$$
\begin{aligned}
\operatorname{var}(\Delta x) &= \operatorname{var}(x_i) + \operatorname{var}(x_j) - 2\operatorname{cov}(x_i,x_j)\\
&= 2\left[\operatorname{var}(x_i) - \operatorname{cov}(x_i,x_j)\right].
\end{aligned}
\tag{E7}
$$

Plugging in the definition of the Pearson correlation we obtain

$$
\begin{aligned}
\operatorname{pearson}(\Delta x,\Delta y) &= \frac{\operatorname{cov}(\Delta x,\Delta y)}{\sqrt{\operatorname{var}(\Delta x)\operatorname{var}(\Delta y)}}\\[4pt]
&= \frac{\operatorname{cov}(x_i,y_i) - \operatorname{cov}(x_i,y_j)}{\sqrt{\operatorname{var}(x_i) - \operatorname{cov}(x_i,x_j)}\,\sqrt{\operatorname{var}(y_i) - \operatorname{cov}(y_i,y_j)}}\\[4pt]
&= \frac{\operatorname{pearson}(x_i,y_i) - \operatorname{pearson}(x_i,y_j)}{\sqrt{1 - \operatorname{pearson}(x_i,x_j)}\,\sqrt{1 - \operatorname{pearson}(y_i,y_j)}},
\end{aligned}
\tag{E8}
$$

where in the last step we divided the numerator and the denominator by $\sqrt{\operatorname{var}(x_i)\operatorname{var}(y_i)}$ and used $\operatorname{var}(x_j) = \operatorname{var}(x_i)$ and $\operatorname{var}(y_j) = \operatorname{var}(y_i)$ for $(i,j)\in\hat{S}$.

Substituting the ranks, and according to the definitions in the previous section, we obtain after re-arrangement

$$
\rho = \hat{\rho}_s + \hat{\rho}_\Delta\sqrt{\left(1-\hat{h}_x\right)\left(1-\hat{h}_y\right)}.
\tag{E9}
$$

Eq. (E9) is an exact finite-sample algebraic identity: every step above uses only sample covariances over $\hat{S}$, where the symmetry of the set makes $\operatorname{var}(x_j) = \operatorname{var}(x_i)$ and $\operatorname{cov}(x_j,y_i) = \operatorname{cov}(x_i,y_j)$ hold exactly. The stochastic interpretation of its two terms in Eqs. (E10)–(E13) is, in contrast, justified at the population level.

Thus, the co-expression can be decomposed into a shared component and an unshared (difference) component. To see this, we write $x_i(t) = x_i^{s}(t) + x_i^{u}(t)$, where $x_i^{s}(t)$ is the contribution of the initial condition inherited from the mother cell, identical in both twins, $x_i^{s}(t) = x_j^{s}(t)$ for $(i,j)\in\hat{S}$, and $x_i^{u}(t)$ is the private contribution that evolved in each cell during $t$ independent of its twin. Similarly, we write $y_i(t) = y_i^{s}(t) + y_i^{u}(t)$. For the ranks, the shared and unshared components are defined directly in rank space using conditional expectations,

$$
r_x^{s} = \mathbb{E}\left[r_x \mid Z\right], \qquad r_x^{u} = r_x - r_x^{s},
$$

and similarly for $y$, where $Z$ represents the shared initial state of the twins. We then obtain

$$
\rho = \underbrace{\frac{\operatorname{cov}\left(r^{s}_{x_i}, r^{s}_{y_i}\right)}{\sqrt{\operatorname{var}\left(r_{x_i}\right)\operatorname{var}\left(r_{y_i}\right)}}}_{=\,\hat{\rho}_s}
+ \underbrace{\frac{\operatorname{cov}\left(r^{u}_{x_i}, r^{u}_{y_i}\right)}{\sqrt{\operatorname{var}\left(r_{x_i}\right)\operatorname{var}\left(r_{y_i}\right)}}}_{=\,\hat{\rho}_\Delta\sqrt{\left(1-\hat{h}_x\right)\left(1-\hat{h}_y\right)}},
\tag{E10}
$$

where we used $\operatorname{cov}(r^{s}_{x_i}, r^{u}_{y_i}) = \operatorname{cov}(r^{s}_{x_i}, r^{u}_{y_j}) = 0$ and $\operatorname{cov}(r^{u}_{x_i}, r^{u}_{y_j}) = 0$ (the same holds when exchanging $x$ and $y$), see the discussion in the next section. Namely, the shared and unshared components enter the co-expression separately, through $\hat{\rho}_s$ and $\hat{\rho}_\Delta$, respectively.

**In TwINFER.** Eq. (E10) shows how twin information discriminates regulatory from non-regulatory correlations:

* **Cell-state heterogeneity.** In a population with two states, where the expression of both genes is low in one state and high in the other, X and Y are correlated even without regulation. For twins, however, $\Delta x \sim \Delta y \sim 0$ shortly after division, regardless of the initial state of the twins, and this remains true later, as long as the elapsed time is much smaller than the dynamic timescale for state transitions. The state is part of the shared initial state $Z$, so it contributes to $\hat{\rho}_s$ and not to $\hat{\rho}_\Delta$. The same applies to a continuous cell-state heterogeneity due to extrinsic noise.
* **Regulation.** If X regulates Y, still $\Delta x \sim \Delta y \sim 0$ shortly after division, and the co-expression is initially shared. However, the regulatory interaction leads to a correlated change in transcript counts within each cell, and $\Delta x$ and $\Delta y$ become more correlated with time: the co-expression shifts to the unshared component.

The weight $\sqrt{(1-\hat{h}_x)(1-\hat{h}_y)}$ sets how much of $\rho$ the unshared component can carry. Right after division, $\hat{h}_x$ and $\hat{h}_y$ are close to 1 and nearly all of $\rho$ is shared; as the trajectories of the twins diverge, the weight grows.

## 3. Interpretation via the law of total covariance

In the previous section we asserted that $\operatorname{cov}(r^{s}_{x_i}, r^{u}_{y_i}) = \operatorname{cov}(r^{s}_{x_i}, r^{u}_{y_j}) = 0$ and $\operatorname{cov}(r^{u}_{x_i}, r^{u}_{y_j}) = 0$. This follows directly from the definition of the shared and unshared parts: the shared part $r^{s}_{x_i}(t)$ is determined by the initial condition $r_{x_i}(0), r_{y_i}(0)$ (and any initial hidden variable that the twins may share), which together form the shared initial state $Z$, and the unshared part is independent between twins given the initial condition. Moreover, by the definition of $r_x^{s}$, the unshared part has zero conditional mean,

$$
\mathbb{E}\left[r_x^{u} \mid Z\right] = 0.
$$

Together with the conditional independence of the twins given $Z$, this justifies the zero-covariance identities in Eqs. (E11)–(E12). By the law of total expectation, averaging first over the dynamics at fixed initial condition, and then over the initial conditions,

$$
\begin{aligned}
\mathbb{E}\left[r^{s}_{x_i}(t)\, r^{u}_{y_j}(t)\right] &= \mathbb{E}\left[r^{s}_{x_i}(t)\, \mathbb{E}\left[r^{u}_{y_j}(t) \mid Z\right]\right]\\
&= \mathbb{E}\left[r^{s}_{x_i}(t)\right] \mathbb{E}\left[r^{u}_{y_j}(t)\right],
\end{aligned}
\tag{E11}
$$

hence

$$
\operatorname{cov}\left(r^{s}_{x_i}, r^{u}_{y_j}\right) = \mathbb{E}\left[r^{s}_{x_i}(t)\, r^{u}_{y_j}(t)\right] - \mathbb{E}\left[r^{s}_{x_i}(t)\right] \mathbb{E}\left[r^{u}_{y_j}(t)\right] = 0.
\tag{E12}
$$

In a similar manner one can show the other two identities.

Using the law of total covariance,

$$
\begin{aligned}
\operatorname{cov}\left(r_{x_i}, r_{y_i}\right) &= \operatorname{cov}\left(\mathbb{E}\left[r_{x_i} \mid Z\right],\, \mathbb{E}\left[r_{y_i} \mid Z\right]\right)\\
&\quad {}+ \mathbb{E}\left[\operatorname{cov}\left(r_{x_i}, r_{y_i} \mid Z\right)\right].
\end{aligned}
\tag{E13}
$$

Plugging in $r_{x_i}(t) = r^{s}_{x_i}(t) + r^{u}_{x_i}(t)$ and $r_{y_i}(t) = r^{s}_{y_i}(t) + r^{u}_{y_i}(t)$ and applying all the above identities, and after division by $\sqrt{\operatorname{var}(r_{x_i})\operatorname{var}(r_{y_i})}$, we obtain Eq. (E9).

**In TwINFER.** The two properties used above, a shared initial state $Z$ and twins that are independent given $Z$, reflect the assumptions of TwINFER. TwINFER relies on the observation that sister cells exhibit greater similarity than random cell pairs drawn from the population, and implicitly assumes that the post-division transcriptome and proteome fully determine the resulting transcriptional steady state (statistically speaking). The framework is therefore best suited for analyzing cells undergoing symmetric partitioning. Higher-order hierarchies of stochastic gene regulation, such as nucleosome positioning, could act as confounding hidden variables if they operate on timescales comparable to transcription factor activity, inducing or masking gene-expression correlation.

In a finite sample, the correlations carry sampling error; TwINFER assesses them with statistical tests (section 4).

## 4. Twin information in TwINFER

TwINFER exploits two kinds of twin information. Within-sample information comes from twin pairs measured in the same sample at the same time point; cross-sample information comes from twins separated between two samples, $S_1$ and $S_2$, collected at time points $t_1 < t_2$, with $t_1$ shortly after the mother cells divided.

**Accounting for cell-state heterogeneity.** Within-sample twins give the twin difference correlations $\hat{\rho}_\Delta(t_1)$ and $\hat{\rho}_\Delta(t_2)$ of Eq. (E5), which, by Eq. (E10), measure the unshared component of the co-expression. Each gene pair goes through a flowchart with three stages:

1. **Stage I.** $\rho$ is tested with a permutation test over scrambled gene-expression profiles. If $\rho$ is below the threshold, there is no significant correlation: a single state without regulation. A value of $\rho$ above the threshold is a necessary condition for regulation, but not a sufficient one, as cell-state heterogeneity can also result in correlations.
2. **Stage II.** $\hat{\rho}_\Delta(t_1)$ is compared, with a z-test, with the distribution of the difference correlation $\rho_\Delta$ over random pairings of cells. Random pairs do not share an initial state, so heterogeneity contributes to $\rho_\Delta$, whereas it cancels in the twin differences. If $\hat{\rho}_\Delta(t_1)$ is not significantly below $\rho_\Delta$, there is a single state, and the correlation is due to regulatory interactions; otherwise, there is cell-state heterogeneity.
3. **Stage III.** A significant increase from $\hat{\rho}_\Delta(t_1)$ to $\hat{\rho}_\Delta(t_2)$ implies heterogeneity *and* regulation; otherwise, heterogeneity alone. Unlike the first two stages, this stage requires repetitions of the experiment or sub-sampling of the data.

| scenario | regulation | cell states | decided at |
|---|---|---|---|
| 1 | no | single | stage I |
| 2 | yes | single | stage II |
| 3 | yes | multiple | stage III |
| 4 | no | multiple | stage III |

Stage II also flags a confounding gene inherited from the mother cells, such as the common regulator of X and Y in a fan-out motif: twins share its initial concentration, and from the perspective of the interaction between X and Y it functions as extrinsic noise.

**Inferring causal relations.** Where the correlation is established solely as regulatory, cross-sample twins resolve the causal relation. TwINFER computes the cross-correlation between $x(t_1)$ and $y(t_2)$, $\hat{\rho}^{\dagger}_{x(t_1)\to y(t_2)}$, for the direction X $\to$ Y, and $\hat{\rho}^{\dagger}_{y(t_1)\to x(t_2)}$ for Y $\to$ X. For a unidirectional interaction only one of them is significant; for a bidirectional interaction both are. The sign of the cross-correlation determines the type of the interaction: positive for activation, negative for repression.

**Lineage-resolved data.** In barcoded datasets, cells may divide several times before they are measured, so the twin pairs are clonal pairs rather than strict sister cells, and all possible cell pairs within each clone are considered. Clonal pairs are significantly more similar to each other than random pairs, and decomposing the co-expression into the two components, shared and unshared by the twins, refines the interpretation of such datasets.
